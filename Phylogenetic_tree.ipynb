```python
from Bio import Phylo
from Bio.Phylo.Newick import Clade



# =====================================================
# INPUT
# =====================================================

taxonomy_file = "taxonomy_fixed.nwk"

timetree_file = "TimeTree_result.nwk"

output_file = "TimeTree_merged.nwk"



# =====================================================
# LOAD TREES
# =====================================================

tax_tree = Phylo.read(
    taxonomy_file,
    "newick"
)


time_tree = Phylo.read(
    timetree_file,
    "newick"
)



# =====================================================
# NORMALIZE NAME
# =====================================================

def normalize_tree(tree):

    for leaf in tree.get_terminals():

        if leaf.name:

            leaf.name = (
                leaf.name
                .replace(" ","_")
                .strip()
            )


normalize_tree(tax_tree)

normalize_tree(time_tree)



# =====================================================
# TAXA SET
# =====================================================

tax_taxa = {

    x.name
    for x in tax_tree.get_terminals()
    if x.name

}



time_taxa = {

    x.name
    for x in time_tree.get_terminals()
    if x.name

}



missing_taxa = tax_taxa - time_taxa



print(
    "Missing taxa:",
    len(missing_taxa)
)


for x in sorted(missing_taxa):

    print(x)



# =====================================================
# FIND NODE OF SPECIES
# =====================================================

def find_species_node(tree, species):


    for leaf in tree.get_terminals():

        if leaf.name == species:

            return leaf


    return None



# =====================================================
# FIND PARENT NODE
# =====================================================

def find_parent(tree, target):


    for clade in tree.find_clades():

        if target in clade.clades:

            return clade


    return None



# =====================================================
# GET EXISTING TIMETREE TAXA INSIDE CLADE
# =====================================================

def existing_species_in_clade(clade):


    result=[]


    if clade is None:

        return result



    for leaf in clade.get_terminals():

        if leaf.name in time_taxa:

            result.append(
                leaf.name
            )


    return result



# =====================================================
# FIND CLOSEST SISTER FROM TAXONOMY
# =====================================================

def find_sister_candidate(
        tree,
        species
):


    node = find_species_node(
        tree,
        species
    )


    if node is None:

        print(
            "Missing in taxonomy:",
            species
        )

        return None



    current = node



    while True:


        parent = find_parent(
            tree,
            current
        )


        if parent is None:

            break



        candidates = (
            existing_species_in_clade(
                parent
            )
        )


        candidates = [
            x for x in candidates
            if x != species
        ]



        if len(candidates)>0:

            return candidates[0]



        current = parent



    return None



# =====================================================
# FIND TIMETREE NODE
# =====================================================

def find_time_node(name):


    return time_tree.find_any(
        name=name
    )



# =====================================================
# INSERT SPECIES
# =====================================================

def insert_sister(
        sister_name,
        missing_name
):


    sister_node = find_time_node(
        sister_name
    )


    if sister_node is None:

        return False



    branch = 0.00001



    new_leaf = Clade(

        name=missing_name,

        branch_length=branch

    )



    parent = Clade(

        branch_length=branch

    )


    parent.clades = [

        sister_node,

        new_leaf

    ]



    def replace(node):


        for i,c in enumerate(node.clades):


            if c is sister_node:

                node.clades[i]=parent

                return True



            if replace(c):

                return True


        return False



    return replace(
        time_tree.root
    )



# =====================================================
# MAIN PIPELINE
# =====================================================

for missing in sorted(missing_taxa):


    print(
        "\nProcessing:",
        missing
    )



    sister = find_sister_candidate(
        tax_tree,
        missing
    )



    if sister is None:


        print(
            "No sister found"
        )

        continue



    print(
        "Nearest TimeTree species:",
        sister
    )



    success = insert_sister(
        sister,
        missing
    )



    if success:

        print(
            "ADDED:",
            missing
        )

    else:

        print(
            "FAILED:",
            missing
        )



# =====================================================
# SAVE
# =====================================================

Phylo.write(
    time_tree,
    output_file,
    "newick"
)


print(
    "\nSaved:",
    output_file
)
```

```python
# Create the phylogenetic tree for Oryzias species + outgroups
from ete3 import NCBITaxa
from ete3.treeview import TreeStyle

# Connect to NCBI taxonomy database
ncbi = NCBITaxa()

# Species list
species = [
    # Oryzias (ricefishes)
    "Oryzias latipes",
    "Oryzias sakaizumii",
    "Oryzias curvinotus",
    "Oryzias luzonensis",
    "Oryzias sinensis",
    "Oryzias mekongensis",
    "Oryzias javanicus",
    "Oryzias melastigma",
    "Oryzias dancena",
    "Oryzias minutillus",
    "Oryzias celebensis",
    "Oryzias marmoratus",
    "Oryzias matanensis",
    "Oryzias nigrimas",
    "Oryzias sarasinorum",
    "Oryzias profundicola",
    "Oryzias setnai",

    # Outgroups
    "Nothobranchius furzeri",
    "Amphiprion ocellaris",
]

# Get NCBI taxonomy IDs
names = ncbi.get_name_translator(species)

taxids = []
for sp in species:
    if sp in names:
        taxids.append(names[sp][0])
    else:
        print(f"Warning: {sp} not found in NCBI taxonomy")

print("TaxIDs:", taxids)

# Build taxonomy tree
tree = ncbi.get_topology(taxids)

# Rename nodes: TaxID -> Latin name
name_dict = ncbi.get_taxid_translator(taxids)

for node in tree.traverse():
    if node.name in name_dict:
        node.name = name_dict[node.name]

# Convert any remaining numeric (TaxID) node names to Latin names
taxids_numeric = [int(node.name) for node in tree.traverse() if node.name.isdigit()]
names_remaining = ncbi.translate_to_names(taxids_numeric)
taxid_to_name = dict(zip(taxids_numeric, names_remaining))

for node in tree.traverse():
    if node.name.isdigit():
        node.name = taxid_to_name[int(node.name)]

# Save tree
tree.write(outfile="Oryzias_phylogeny.nwk")

print("Done: Oryzias_phylogeny.nwk")

# Optional: quick visualization in text
print(tree)
```
